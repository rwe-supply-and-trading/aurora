## ERA5 Raw Data

Surface Level Variables

```python
import zarr
import kafou_arraylake as arraylake
import xarray as xr

source_repo = "rwe/era5-0p25-6h-nonprod-ohio"
source_branch = "main"

client = arraylake.Client()

repo = client.get_repo(source_repo)
session = repo.readonly_session(source_branch)

sfc_ds = xr.open_zarr(session.store, group="surface", consolidated=False)  # no zarr_format override
sfc_ds
```

Pressure Level Variables

```python
pl_ds = xr.open_zarr(session.store, group="pressure_level", consolidated=False)  # no zarr_format override
pl_ds
```

Invariant Variables

```python
inv_ds = xr.open_zarr(session.store, group="invariant", consolidated=False)  # no zarr_format override
inv_ds
```
